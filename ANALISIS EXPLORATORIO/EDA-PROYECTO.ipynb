## ¿Como calcular la matriz de tiempos de desplazamiento? 

1. La red vial, ¿de dónde sale el grafo?

La fuente estándar es OpenStreetMap (OSM). Se descarga el extracto de la ciudad y ya tendriamos calle con su geometría, sentido, velocidad implícita por tipo de vía, y restricciones de giro. Hay dos formas de usarlo:

a. osmnx (Python, de Geoff Boeing): descarga, construye y analiza el grafo vial directo desde la API de Overpass. Es lo más rápido para empezar a explorar con unas cuentas coordenadas. Por ahora creo que debemos simular coordenadas para las estaciones de donde partiran las ambulancias. 
b. Archivo PBF: para motores de ruteo dedicados bajas el extracto de Geofabrik o BBBike y lo se alimenta al motor.

2. Motores de ruteo 

| Motor | Tipo | Fortaleza | Limitación |
|---|---|---|---|
| **OSRM** | Self-hosted o demo API | Muy rápido, tabla de distancias N×N nativa | Solo auto/bici/pie, no tráfico en tiempo real |
| **Valhalla** | Self-hosted | Isócronas, matrices, más modos | Un poco más pesado de montar |
| **GraphHopper** | Self-hosted o cloud | Flexible, custom profiles | Versión open tiene menos features |
| **Google Directions API** | Cloud | Tráfico real, muy preciso | Cuesta plata y tiene rate limits |
| **openrouteservice** | API pública o self-hosted | Isócronas y matrices gratis (con límites) | Rate limit en la API pública |


Para este proyecto **OSRM** es parece ser la mejor opción. Es gratis y tiene el endpoint /table da la matriz de tiempos de viaje de N orígenes a M destinos en una sola llamada.

**OSMR** se puede arrancar con docker o desde python. Esta ultima opcion tiene  un desempeño limitado si el grafo crece mucho (+100 puntos) seguramente nuestro grafo tendrá mucho más de 100 nodos. Por ahora, para ir haciendo pruebas podemos correr todo en python y luego vemos como lo optimizamos. 

3. La matriz de tiempos/distancias — lo que alimentaria el modelo de optimización que queremos hacer

Si se tiene K estaciones (De donde saldrán las ambulancias) + N emergencias, necesitamos una matriz (K+N) × (K+N) con el tiempo de viaje entre cada par de puntos por la red vial de la ciudad.

4. Simulación - Casos de contingencia 

- Usando solo python (osmnx + NetworkX): Como el grafo es un objeto de Python que vive en la memoria, simplemente se remueven aristas o le sube el peso a infinito (una arista representa una calle, avenida, carrera,etc).

- Si se usa OSMR con docker: OSRM precompila el grafo, entonces no se puede modificarlo en caliente. Hay tres opciones:

La más práctica es editar el archivo OSM antes de compilar. Descargar el PBF, y convertirlo a XML, filtrar las vías que quieremos cerrar y luego recompilar. Es más rapido pero se requiere más trabajo. 



| Necesidad | osmnx/NetworkX | OSRM |
|---|---|---|
| Cerrar una calle | `.remove_edges_from()` instantáneo | Recompilar el PBF (~minutos) |
| Penalizar (congestión) | Modificar `travel_time` en memoria | Editar perfil Lua + recompilar |
| Muchos escenarios | Muy flexible | Tedioso |
| Matriz 500×500 | Lento (~minutos) | Milisegundos |

### Implementacion rapida (para pruebas) de osmnx + networkX en python

```python
import osmnx as ox
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
import warnings
import sklearn
warnings.filterwarnings("ignore")
```

```python
# Configuración de osmnx
ox.settings.use_cache = True      # Cachea descargas para no repetir
ox.settings.log_console = False
```

```python
#OPCIÓN A: Por nombre de ciudad ---
G = ox.graph_from_place("Montgomery County, Pennsylvania, USA", network_type="drive")
```

```python
print(f"Nodos: {G.number_of_nodes():,}")
print(f"Aristas: {G.number_of_edges():,}")
```

```python
# Agregar velocidades y tiempos de viaje
# osmnx asigna velocidades según el tipo de vía (highway tag de OSM)
# y calcula travel_time en segundos para cada arista
G = ox.add_edge_speeds(G)
G = ox.add_edge_travel_times(G)

# Verificar que se asignaron
ejemplo = list(G.edges(data=True))[0]
print(f"Ejemplo de arista:")
print(f"  Longitud: {ejemplo[2].get('length', '?'):.0f} m")
print(f"  Velocidad: {ejemplo[2].get('speed_kph', '?')} km/h")
print(f"  Tiempo: {ejemplo[2].get('travel_time', '?'):.1f} seg")
```

```python
#Definir puntos (estaciones y emergencias)
# Ejemplo: dos hopitales/centros de emergencias 
# y varias emergencias (salen del dataset que tenemos)


estaciones = pd.DataFrame({
    "id": ["E1", "E2"],
    "Desc": ["Jefferson Einstein Montgomery Hospital", "Montgomery County Emergency Service"],
    "Lat": [40.152678, 40.136960],
    "Lon": [-75.341897
, -75.353025
],
})
#tomamos las primeras 5 filas del dataframe ya creado para el EDA
emergencias = df.loc[1:6, ["lat","lng","title"]]
emergencias.columns = ["Lat", "Lon", "Desc"]

emergencias.head()
```

```python
estaciones.head()
```

```python
# Punto de retorno al final del turno
punto_retorno = {"Lat": 40.140046, "Lon": -75.311119,"Desc":"Centro de Recolección de Residuos"}
# Combinar todos los puntos para la matriz
todos = pd.concat([
    estaciones[["Desc", "Lat", "Lon"]],
    emergencias[["Desc", "Lat", "Lon"]],
    pd.DataFrame([punto_retorno])
], ignore_index=True)

print(todos)
```

```python
# Cada coordenada se asocia al nodo más cercano del grafo

todos["nodo_osm"] = todos.apply(
    lambda row: ox.nearest_nodes(G, X=row["Lon"], Y=row["Lat"]),
    axis=1
)
print(todos)
```
